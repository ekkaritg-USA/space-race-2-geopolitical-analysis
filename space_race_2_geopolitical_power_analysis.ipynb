##**Step 3: Building the Space Power Index (SPI)**

**Step 3a: Calculate Component Metrics**

````python
print("🏆 STEP 3: Building the Space Power Index")
print("="*60)

# We'll calculate SPI for different time periods to show evolution
# Let's start with the CURRENT era (2016-2022)

current_era = df[df['Year'] >= 2016].copy()

print(f"\nAnalyzing current era: 2016-2022")
print(f"Total missions in this period: {len(current_era):,}")

# Calculate metrics by country for current era
country_metrics = current_era.groupby('Country').agg({
    'Mission': 'count',  # Launch volume
    'Success': 'sum',    # Successful launches
    'Rocket': 'nunique'  # Technology diversity
}).rename(columns={
    'Mission': 'Launch_Count',
    'Success': 'Successful_Launches',
    'Rocket': 'Unique_Rockets'
})

# Calculate success rate
country_metrics['Success_Rate'] = (
    country_metrics['Successful_Launches'] / country_metrics['Launch_Count'] * 100
).round(2)

# Calculate launches per year (momentum)
years_in_period = 7  # 2016-2022 is 7 years
country_metrics['Launches_Per_Year'] = (
    country_metrics['Launch_Count'] / years_in_period
).round(2)

# Only keep countries with at least 5 launches (filter noise)
country_metrics = country_metrics[country_metrics['Launch_Count'] >= 5]

print(f"\nCountries with 5+ launches (2016-2022): {len(country_metrics)}")
print("\nRaw metrics:")
print(country_metrics.sort_values('Launch_Count', ascending=False))
````

**Step 3b: Normalize All Metrics (0-100 Scale)**

````python
# Normalization function (min-max scaling to 0-100)
def normalize_to_100(series):
    """Normalize a series to 0-100 scale"""
    min_val = series.min()
    max_val = series.max()
    if max_val == min_val:
        return pd.Series([50] * len(series), index=series.index)
    return ((series - min_val) / (max_val - min_val) * 100).round(2)

# Normalize each component
country_metrics['Volume_Score'] = normalize_to_100(country_metrics['Launch_Count'])
country_metrics['Success_Score'] = country_metrics['Success_Rate']  # Already 0-100
country_metrics['Diversity_Score'] = normalize_to_100(country_metrics['Unique_Rockets'])
country_metrics['Momentum_Score'] = normalize_to_100(country_metrics['Launches_Per_Year'])

print("\n📊 Normalized Scores (0-100):")
print(country_metrics[['Volume_Score', 'Success_Score', 'Diversity_Score', 'Momentum_Score']].sort_values('Volume_Score', ascending=False))
````

**Step 3c: Calculate Space Power Index**

````python
# Space Power Index Formula:
# SPI = (Volume × 0.25) + (Success × 0.30) + (Diversity × 0.20) + (Momentum × 0.25)

weights = {
    'Volume': 0.25,
    'Success': 0.30,
    'Diversity': 0.20,
    'Momentum': 0.25
}

country_metrics['Space_Power_Index'] = (
    country_metrics['Volume_Score'] * weights['Volume'] +
    country_metrics['Success_Score'] * weights['Success'] +
    country_metrics['Diversity_Score'] * weights['Diversity'] +
    country_metrics['Momentum_Score'] * weights['Momentum']
).round(2)

# Rank countries
country_metrics['Rank'] = country_metrics['Space_Power_Index'].rank(ascending=False, method='min').astype(int)

# Sort by rank
country_metrics = country_metrics.sort_values('Space_Power_Index', ascending=False)

print("\n" + "="*60)
print("🏆 SPACE POWER INDEX - CURRENT ERA (2016-2022)")
print("="*60)
print("\nTop 10 Space Powers:")
print(country_metrics[['Rank', 'Space_Power_Index', 'Launch_Count', 'Success_Rate', 'Unique_Rockets', 'Launches_Per_Year']].head(10))

print("\n💡 Index Interpretation:")
print("  • 80-100: Dominant Space Power")
print("  • 60-79: Major Space Power")
print("  • 40-59: Emerging Space Power")
print("  • 20-39: Developing Space Capability")
print("  • 0-19: Limited Space Activity")
````

**Step 3d: Calculate Historical SPI (By Decade)**

````python
# Let's calculate SPI for each decade to show evolution
print("\n📈 Calculating historical Space Power Index by decade...")

decades = [1960, 1970, 1980, 1990, 2000, 2010, 2020]
historical_spi = []

for decade in decades:
    # Get data for this decade
    decade_data = df[(df['Year'] >= decade) & (df['Year'] < decade + 10)].copy()

    if len(decade_data) == 0:
        continue

    # Calculate metrics
    decade_metrics = decade_data.groupby('Country').agg({
        'Mission': 'count',
        'Success': 'sum',
        'Rocket': 'nunique'
    })

    decade_metrics['Success_Rate'] = (
        decade_metrics['Success'] / decade_metrics['Mission'] * 100
    )

    decade_metrics['Launches_Per_Year'] = decade_metrics['Mission'] / 10

    # Filter countries with at least 5 launches
    decade_metrics = decade_metrics[decade_metrics['Mission'] >= 5]

    if len(decade_metrics) == 0:
        continue

    # Normalize
    decade_metrics['Volume_Score'] = normalize_to_100(decade_metrics['Mission'])
    decade_metrics['Success_Score'] = decade_metrics['Success_Rate']
    decade_metrics['Diversity_Score'] = normalize_to_100(decade_metrics['Rocket'])
    decade_metrics['Momentum_Score'] = normalize_to_100(decade_metrics['Launches_Per_Year'])

    # Calculate SPI
    decade_metrics['Space_Power_Index'] = (
        decade_metrics['Volume_Score'] * 0.25 +
        decade_metrics['Success_Score'] * 0.30 +
        decade_metrics['Diversity_Score'] * 0.20 +
        decade_metrics['Momentum_Score'] * 0.25
    ).round(2)

    decade_metrics['Decade'] = f"{decade}s"
    decade_metrics['Rank'] = decade_metrics['Space_Power_Index'].rank(ascending=False, method='min').astype(int)

    historical_spi.append(decade_metrics.reset_index())

# Combine all decades
historical_spi_df = pd.concat(historical_spi, ignore_index=True)

print("\n✅ Historical SPI calculated for all decades!")
print(f"\nSample - Top 3 in each decade:")

for decade in decades:
    decade_name = f"{decade}s"
    decade_top = historical_spi_df[historical_spi_df['Decade'] == decade_name].nsmallest(3, 'Rank')
    if len(decade_top) > 0:
        print(f"\n{decade_name}:")
        for idx, row in decade_top.iterrows():
            print(f"  {int(row['Rank'])}. {row['Country']}: {row['Space_Power_Index']:.1f}")
````

**Step 5b: Emerging Power Analysis**

````python
# Identify emerging powers by growth rate
# Compare 2010-2015 vs 2016-2022

early_modern = df[(df['Year'] >= 2010) & (df['Year'] < 2016)]
recent = df[df['Year'] >= 2016]

early_counts = early_modern.groupby('Country')['Mission'].count()
recent_counts = recent.groupby('Country')['Mission'].count()

# Calculate growth
growth_data = []
for country in recent_counts.index:
    if country in early_counts.index and early_counts[country] >= 5:  # At least 5 launches in early period
        early = early_counts[country]
        recent_val = recent_counts[country]
        growth_pct = ((recent_val - early) / early * 100)
        growth_data.append({
            'Country': country,
            'Launches_2010_2015': early,
            'Launches_2016_2022': recent_val,
            'Growth_%': growth_pct,
            'Color': COUNTRY_COLORS.get(country, '#808080')
        })

growth_df = pd.DataFrame(growth_data).sort_values('Growth_%', ascending=False)

# Filter to significant players (at least 10 launches total)
growth_df = growth_df[growth_df['Launches_2016_2022'] >= 10]

print("\n📈 EMERGING POWERS ANALYSIS:")
print(growth_df)

# Visualize emerging powers
fig6 = go.Figure()

fig6.add_trace(go.Bar(
    x=growth_df['Country'],
    y=growth_df['Growth_%'],
    marker=dict(
        color=growth_df['Color'],
        line=dict(color='white', width=1.5)
    ),
    text=growth_df['Growth_%'].round(0),
    texttemplate='%{text}%',
    textposition='outside',
    textfont=dict(size=12, family='Arial Black'),
    hovertemplate='<b>%{x}</b><br>' +
                  '2010-2015: %{customdata[0]} launches<br>' +
                  '2016-2022: %{customdata[1]} launches<br>' +
                  'Growth: %{y:.0f}%<br>' +
                  '<extra></extra>',
    customdata=growth_df[['Launches_2010_2015', 'Launches_2016_2022']].values
))

fig6.update_layout(
    title={
        'text': '<b>Emerging Space Powers: Growth Rate (2010-15 vs 2016-22)</b><br>' +
                '<sub>Which nations are accelerating their space programs?</sub>',
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 22, 'family': 'Arial Black'}
    },
    xaxis_title='Country',
    yaxis_title='Growth Rate (%)',
    height=600,
    width=1200,
    template='plotly_white',
    showlegend=False
)

# Add reference line at 0%
fig6.add_hline(
    y=0,
    line_dash="dash",
    line_color="gray",
    annotation_text="No Growth",
    annotation_position="right"
)

fig6.show()
fig6.write_html('viz6_emerging_powers_growth.html')
print("✅ Viz 6: Emerging Powers Growth Analysis")
````

**Step 5c: Prediction (Simple Linear Projection)**

````python
# Simple prediction: Linear trend for USA and China
# Based on 2016-2022 data

usa_annual = df[(df['Country'] == 'USA') & (df['Year'] >= 2016)].groupby('Year')['Mission'].count()
china_annual = df[(df['Country'] == 'China') & (df['Year'] >= 2016)].groupby('Year')['Mission'].count()

# Calculate average annual launches
usa_avg = usa_annual.mean()
china_avg = china_annual.mean()

# Project to 2030
projection_years = list(range(2023, 2031))
usa_projection = [usa_avg * 1.05 ** (year - 2022) for year in projection_years]  # 5% annual growth
china_projection = [china_avg * 1.08 ** (year - 2022) for year in projection_years]  # 8% annual growth

# Create projection visualization
fig7 = go.Figure()

# Historical data
fig7.add_trace(go.Scatter(
    x=usa_annual.index,
    y=usa_annual.values,
    mode='lines+markers',
    name='USA (Historical)',
    line=dict(color=COUNTRY_COLORS['USA'], width=4),
    marker=dict(size=10)
))

fig7.add_trace(go.Scatter(
    x=china_annual.index,
    y=china_annual.values,
    mode='lines+markers',
    name='China (Historical)',
    line=dict(color=COUNTRY_COLORS['China'], width=4),
    marker=dict(size=10)
))

# Projections
fig7.add_trace(go.Scatter(
    x=projection_years,
    y=usa_projection,
    mode='lines+markers',
    name='USA (Projected)',
    line=dict(color=COUNTRY_COLORS['USA'], width=3, dash='dash'),
    marker=dict(size=8, symbol='diamond')
))

fig7.add_trace(go.Scatter(
    x=projection_years,
    y=china_projection,
    mode='lines+markers',
    name='China (Projected)',
    line=dict(color=COUNTRY_COLORS['China'], width=3, dash='dash'),
    marker=dict(size=8, symbol='diamond')
))

# Add vertical line at 2022 (current data cutoff)
fig7.add_vline(
    x=2022.5,
    line_dash="dot",
    line_color="gray",
    annotation_text="Data Ends | Projection Begins",
    annotation_position="top"
)

fig7.update_layout(
    title={
        'text': '<b>USA vs China: Launch Activity Projection to 2030</b><br>' +
                '<sub>Based on 2016-2022 trends (USA: 5% growth, China: 8% growth assumptions)</sub>',
        'x': 0.5,
        'xanchor': 'center',
        'font': {'size': 22, 'family': 'Arial Black'}
    },
    xaxis_title='Year',
    yaxis_title='Annual Launches',
    height=600,
    width=1300,
    template='plotly_white',
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=-0.2,
        xanchor="center",
        x=0.5,
        font=dict(size=12),
        bgcolor='rgba(255, 255, 255, 0.9)',
        bordercolor='black',
        borderwidth=1
    ),
    hovermode='x unified'
)

fig7.show()
fig7.write_html('viz7_2030_projections.html')
print("✅ Viz 7: 2030 Predictions")

# Print key findings
print("\n🔮 2030 PROJECTION KEY FINDINGS:")
print(f"  • USA 2030 (projected): ~{int(usa_projection[-1])} launches/year")
print(f"  • China 2030 (projected): ~{int(china_projection[-1])} launches/year")
if china_projection[-1] > usa_projection[-1]:
    print(f"  • China projected to lead by ~{int(china_projection[-1] - usa_projection[-1])} launches/year")
else:
    print(f"  • USA projected to maintain lead")
````
